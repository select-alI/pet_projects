---

### Проверим, одинаков ли размер выборок

```python
SRM_df = res_df.group.value_counts().reset_index().rename(columns={'count': 'observed'})
SRM_df['expected'] = SRM_df.observed.sum() / 2
SRM_df
```

**Гипотезы:**

- H₀: Размер выборок одинаковый

- H₁: Размер выборок разный

*alpha* = 0.01

```python
from scipy.stats import chisquare
alpha = 0.01

chi2_stat, p_value = chisquare(f_obs=SRM_df.observed, f_exp=SRM_df.expected)
```

```python
print(f"χ² статистика: {chi2_stat:.6f}")

if p_value < alpha:
    print(f'p-value: {p_value:.3f} \nОтклоняем нулевую гипотезу')
elif p_value >= alpha:
    print(f'p-value: {p_value:.3f} \nНет оснований отклонить нулевую гипотезу')
```

Делаем вывод, что размер групп *одинаковый*

Проверим, есть ли статистически значимые различия в конверсии

### Z тест для пропорций:

**Сформулируем гипотезы**

* H0: конверсия в контрольной группе равна конверсии в экспериментальной 
* H1: конверсии различаются


**Выберем уровень значимости a = 0.05**

```python
a = 0.05
```

**Рассчитаем число успешых *k* и общее число *n* испытаний в обеих группах**

```python
count_df = res_df.groupby('group').agg(k=('converted', 'sum'),
                                       n=('converted', 'count'))
count_df
```

```python
k_control = count_df.loc['control', 'k']
k_treatment = count_df.loc['treatment', 'k']
n_control = count_df.loc['control', 'n']
n_treatment = count_df.loc['treatment', 'n']
```

**Рассчитаем Z статистику**

```python
from scipy import stats
#cr - conversion rate

cr_control = conv.loc[0, 'conversion']  # конверсия в контрольной группе
cr_treatment = conv.loc[1, 'conversion']  # Конверсия в тестовой группе

# Рассчитаем Z статистику
Z = (cr_treatment - cr_control) / np.sqrt(cr_control*(1-cr_control) / n_control + cr_treatment * (1-cr_treatment) / n_treatment)
```

**Найдем p_value**

```python
# cdf(x): вероятность того, что случайная величина примет значение меньшее или равное x

p_value = 2 * (1 - stats.norm.cdf(np.abs(Z)))
```

```python
print(f'Уровень значимости: {a}')
print(f'Z-статистика: {Z:.3f}')
if p_value < 0.05:
    print(f'p-value: {p_value:.3f}. \nОтклоняем нулевую гипотезу')
elif p_value >= 0.05:
    print(f'p-value: {p_value:.3f}. \nНет оснований отклонить нулевую гипотезу')
else:
    print('Проверьте корректность данных')
```

- Проверим результат

```python
from statsmodels.stats.proportion import proportions_ztest

stat, p_value = proportions_ztest([k_treatment, k_control], [n_treatment, n_control])
print(f"Z-статистика = {stat:.3f}\np-value = {p_value:.3f}")
```

### Построим 95% доверительный интервал для полученных результатов

```python
control_cr, treatment_cr = conv.conversion
diff = treatment_cr - control_cr
conv
```

```python
SE = np.sqrt(control_cr * (1 - control_cr) / n_control + treatment_cr * (1 - treatment_cr) / n_treatment)

Z_score = stats.norm.ppf(1 - 0.05 / 2)
```

```python
lower = diff - Z_score * SE
upper = diff + Z_score * SE
print(f'Нижняя граница: {lower}')
print(f'Среднее Значение:{diff}')
print(f'Верхняя граница: {upper}')
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 1.5))

# 1. Линия интервала
plt.hlines(0, lower, upper, colors='blue', linewidth=6, alpha=0.7)

# 2. Точка оценки
plt.plot(diff, 0, 'ro', markersize=10)

# 3. Линия нуля
plt.axvline(0, color='black', linestyle=':', linewidth=1)

# Настройки
plt.xlim(-0.006, 0.003)
plt.ylim(-0.7, 0.7)
plt.yticks([])
plt.xticks([-0.004, -0.002, 0, 0.002], ['-0.4%', '-0.2%', '0%', '+0.2%'])
plt.xlabel('Разница в конверсии')
plt.grid(True, alpha=0.2, axis='x')

plt.tight_layout()
plt.show()
```

Таким образом, если эффект и есть в дейтвительности, то он незначительный (и скорее всего отрицательный). Надо оставить предыдущую версию страницы, эксперимент продолжать не стоит.

### * Проанализируем время сесси

```python
res_df.head(3)
```

```python
print(res_df.timestamp.dtypes)
```

```python
# переведем вреям в секунды
def to_sec(time):
    minutes, sec = time.split(':')
    return int(minutes) * 60 + float(sec)

res_df['sec'] = res_df.timestamp.apply(to_sec)
```

```python
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)
```

```python
plt.figure(figsize=(7, 3.5))
sns.histplot(res_df, x='sec', bins=20)
plt.xlabel('seconds')
plt.title('График распеделения людей по времени сессии')
plt.show()
```

```python
(res_df.sec / 60).max()
```

```python
intervals = [
    "до 10 минут",
    "10-20 минут",
    "20-30 минут", 
    "30-40 минут",
    "40-50 минут",
    "50-60 минут"
]
res_df['time_intervals'] = pd.cut(res_df['sec'] / 60, bins=6, labels=intervals)
```

```python
res_df.head()
```

рассчитаем конверсию в разных временных интервалах

```python
intervals_df = res_df.groupby(['group', 'time_intervals'], as_index=False).agg(conversion=('converted', 'mean'))
```

```python
plt.figure(figsize=(8, 4))
sns.lineplot(intervals_df, x='time_intervals', y='conversion', hue='group', marker='o')

plt.ylim(0.115, 0.123)
plt.title('График зависимости конверсии от времени сесси')
plt.show()
```
